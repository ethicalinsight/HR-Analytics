#Dataset Description
This dataset is a fictional data set created by IBM data scientists to uncover the factors that lead to employee attrition. It represents human resources (HR) metrics, employee demographics, job roles, satisfaction scores, and performance records for **1,470 employees** within an organization.

The primary goal of analyzing this dataset is to identify key drivers of employee turnover (attrition) and build predictive models to foresee which employees are most likely to leave the company.

---

## Column Descriptions

| Column Name | Data Type | Description |
| :--- | :--- | :--- |
| **Age** | Integer | The age of the employee. |
| **Attrition** | Categorical | Indicates whether the employee left the organization (`Yes` or `No`). |
| **BusinessTravel** | Categorical | Frequency of business travel (`Non-Travel`, `Travel_Rarely`, `Travel_Frequently`). |
| **DailyRate** | Integer | The daily monetary compensation rate for the employee. |
| **Department** | Categorical | The department in which the employee works (e.g., `Sales`, `Research & Development`, `Human Resources`). |
| **DistanceFromHome** | Integer | The distance in miles between the employee's home and workplace. |
| **Education** | Ordinal (1–5) | The education level of the employee (1: Below College, 2: College, 3: Bachelor, 4: Master, 5: Doctor). |
| **EducationField** | Categorical | The field of study of the employee's education (e.g., `Life Sciences`, `Medical`, `Marketing`, `Technical Degree`). |
| **EmployeeCount** | Integer | Constant value of 1 for each record (used for administrative counting). |
| **EmployeeNumber** | Integer | Unique identification number assigned to each employee. |
| **EnvironmentSatisfaction** | Ordinal (1–4) | The employee's satisfaction level with their physical and social working environment (1: Low, 2: Medium, 3: High, 4: Very High). |
| **Gender** | Categorical | The gender of the employee (`Male`, `Female`). |
| **HourlyRate** | Integer | The hourly wage or monetary rate of the employee. |
| **JobInvolvement** | Ordinal (1–4) | The degree of involvement or engagement the employee has with their job (1: Low, 2: Medium, 3: High, 4: Very High). |
| **JobLevel** | Integer | The hierarchical level of the employee's job role within the company (ranging from 1 to 5). |
| **JobRole** | Categorical | The specific job title or position held by the employee (e.g., `Sales Executive`, `Research Scientist`, `Laboratory Technician`). |
| **JobSatisfaction** | Ordinal (1–4) | The employee's satisfaction level with their overall job (1: Low, 2: Medium, 3: High, 4: Very High). |
| **MaritalStatus** | Categorical | The marital status of the employee (`Single`, `Married`, `Divorced`). |
| **MonthlyIncome** | Integer | The monthly salary or compensation earned by the employee. |
| **MonthlyRate** | Integer | The monthly rate used to calculate compensation and benefits. |
| **NumCompaniesWorked** | Integer | Total number of different companies or organizations the employee has worked for prior to this one. |
| **Over18** | Categorical | Indicates whether the employee is over 18 years of age (`Y` or `N`). |
| **OverTime** | Categorical | Indicates whether the employee works overtime hours (`Yes` or `No`). |
| **PercentSalaryHike** | Integer | The percentage increase in salary granted in the most recent performance review. |
| **PerformanceRating** | Ordinal (1–4) | The employee's performance evaluation rating (3: Excellent, 4: Outstanding, etc.). |
| **RelationshipSatisfaction** | Ordinal (1–4) | The employee's satisfaction with workplace relationships and peer interactions (1: Low, 2: Medium, 3: High, 4: Very High). |
| **StandardHours** | Integer | Standard working hours per week (constant value of 80 in this dataset). |
| **StockOptionLevel** | Integer | The level of stock options granted to the employee (ranging from 0 to 3). |
| **TotalWorkingYears** | Integer | The total cumulative number of years the employee has worked across their career. |
| **TrainingTimesLastYear** | Integer | The number of training sessions or programs attended by the employee over the past year. |
| **WorkLifeBalance** | Ordinal (1–4) | The employee's assessment of their work-life balance (1: Bad, 2: Good, 3: Better, 4: Best). |
| **YearsAtCompany** | Integer | The number of years the employee has spent working at the current company. |
| **YearsInCurrentRole** | Integer | The number of years the employee has spent working in their current specific job role. |
| **YearsSinceLastPromotion** | Integer | The number of years that have passed since the employee's last promotion. |
| **YearsWithCurrManager** | Integer | The number of years the employee has worked under their current direct manager. |

###Data Cleaning & Formatting

```python
import pandas as pd
import numpy as np

# 2. Re-apply your successful Feature Engineering steps (TenureRatio, PromotionLag, AgeGroup, SalaryGroup)
df['TenureRatio'] = np.where(
    df['TotalWorkingYears'] == 0,
    0,
    df['YearsAtCompany'] / df['TotalWorkingYears']
)

df['PromotionLag'] = df['YearsSinceLastPromotion']

age_bins = [18, 25, 35, 45, 55, 60]
age_labels = ['18-25', '26-35', '36-45', '46-55', '56-60']
df['AgeGroup'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels, right=False)

salary_bins = [0, 2500, 5000, 7500, 10000, 15000, np.inf]
salary_labels = ['<2.5K', '2.5K-5K', '5K-7.5K', '7.5K-10K', '10K-15K', '>15K']
df['SalaryGroup'] = pd.cut(df['MonthlyIncome'], bins=salary_bins, labels=salary_labels, right=False)

# 3. Update your column lists
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()

print("Dataset successfully reset and feature-engineered!")
print("Attrition unique values (should be 'Yes'/'No'):", df['Attrition'].unique())
print("Updated numerical_cols count:", len(numerical_cols))
print("Updated categorical_cols:", categorical_cols)
```

###Feature Engineering

```python
# Create TenureRatio: YearsAtCompany divided by TotalWorkingYears
df['TenureRatio'] = np.where(
    df['TotalWorkingYears'] == 0,
    0, # Handle division by zero, assuming 0 if no total working years
    df['YearsAtCompany'] / df['TotalWorkingYears']
)

# Create PromotionLag: Time elapsed since the last promotion
df['PromotionLag'] = df['YearsSinceLastPromotion']

# Create Age Brackets
age_bins = [18, 25, 35, 45, 55, 60]
age_labels = ['18-25', '26-35', '36-45', '46-55', '56-60']
df['AgeGroup'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels, right=False)

# Create Salary Brackets (using MonthlyIncome)
salary_bins = [0, 2500, 5000, 7500, 10000, 15000, np.inf] # np.inf for the upper bound of the last bin
salary_labels = ['<2.5K', '2.5K-5K', '5K-7.5K', '7.5K-10K', '10K-15K', '>15K']
df['SalaryGroup'] = pd.cut(df['MonthlyIncome'], bins=salary_bins, labels=salary_labels, right=False)

# Update numerical and categorical column lists after adding new features
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()

print("New features created: 'TenureRatio', 'PromotionLag', 'AgeGroup', 'SalaryGroup'.")
print("Updated numerical_cols:", numerical_cols)
print("Updated categorical_cols:", categorical_cols)
```
